### 📌 الطريقة الثانية: Python Keep-Alive (من داخل Notebook)

شغّل الخلية التالية:

````python
from IPython.display import display, HTML

# JavaScript لمنع قطع الاتصال
keep_alive_js = """
<script>
function ClickConnect(){
  console.log("⏰ Keep-Alive: Clicking connect button...");
  var button = document.querySelector("colab-toolbar-button#connect");
  if(button) button.click();
}
setInterval(ClickConnect, 60000); // كل دقيقة
console.log("✅ Keep-Alive Started!");
</script>

<div style='padding: 15px; background: #0f9d58; color: white; border-radius: 8px; text-align: center;'>
  <h3>✅ Keep-Alive مفعّل!</h3>
  <p>Colab لن يفصل الاتصال تلقائياً</p>
</div>
"""

display(HTML(keep_alive_js))
````

## 📊 الخطوة 6: مراقبة حالة البوت

استخدم الخلية التالية لفحص حالة البوت في أي وقت:

````python
import requests
import json

# الحصول على Ngrok URL الحالي
tunnels = ngrok.get_tunnels()
if tunnels:
    public_url = tunnels[0].public_url
    print(f"🌐 Current Ngrok URL: {public_url}")
    
    # فحص صحة البوت
    try:
        response = requests.get(f"{public_url}/health", timeout=5)
        if response.status_code == 200:
            data = response.json()
            print("\n✅ البوت شغال بنجاح!")
            print(f"   الحالة: {data.get('status')}")
            print(f"   الوقت: {data.get('timestamp')}")
        else:
            print(f"\n⚠️ البوت يرد لكن في مشكلة (Status: {response.status_code})")
    except Exception as e:
        print(f"\n❌ البوت مش شغال: {e}")
else:
    print("❌ مفيش Ngrok tunnel شغال")
````

## 📈 الخطوة 7: عرض إحصائيات البوت

````python
import json
from datetime import datetime

# قراءة بيانات المستخدمين
try:
    with open('users_data.json', 'r', encoding='utf-8') as f:
        users_data = json.load(f)
    
    print("📊 إحصائيات البوت:")
    print("=" * 50)
    print(f"👥 عدد المستخدمين: {len(users_data)}")
    
    # حساب عدد المشتركين النشطين
    active_subs = sum(1 for user in users_data.values() 
                     if user.get('subscription', {}).get('active', False))
    print(f"⭐ المشتركين النشطين: {active_subs}")
    
    # حساب إجمالي الأسئلة
    total_questions = sum(user.get('questions_count', 0) for user in users_data.values())
    print(f"❓ إجمالي الأسئلة: {total_questions}")
    
    print("=" * 50)
    
except FileNotFoundError:
    print("❌ ملف users_data.json غير موجود")
except json.JSONDecodeError:
    print("❌ خطأ في قراءة ملف users_data.json")
````
